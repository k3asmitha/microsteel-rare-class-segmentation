```python
import os
import shutil

# Existing completed Colab run
source_run_id = "II-AUN-P-COLAB"

# Official experiment ID
run_id = "II-AUN-P"

drive_dir = f"/content/drive/MyDrive/MicroSteel/experiment_results/{run_id}"
os.makedirs(drive_dir, exist_ok=True)

# Copy training run
shutil.copytree(
    f"/content/microsteel_project/outputs/training_runs/{source_run_id}",
    f"{drive_dir}/training_run",
    dirs_exist_ok=True
)

# Copy evaluation results
result_dir = (
    "/content/microsteel_project/outputs/results/"
    "Type_II/AttentionUNet/patch"
)

if os.path.exists(result_dir):
    shutil.copytree(
        result_dir,
        f"{drive_dir}/results",
        dirs_exist_ok=True
    )

print(f"\n✓ {run_id} saved")
print(f"✓ Backup saved to: {drive_dir}")
```
